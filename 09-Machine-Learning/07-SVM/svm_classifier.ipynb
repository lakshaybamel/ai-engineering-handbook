# 🧠 Support Vector Machine (SVM) Classifier

## 📌 Objective
Build an SVM classifier using Iris dataset and understand the effect of kernels and hyperparameters.

## 📊 Dataset
Iris Dataset (sklearn)

## ⚙️ Workflow
1. Load Data  
2. Preprocess Data  
3. Train Model  
4. Try Different Kernels  
5. Tune Hyperparameters  
6. Evaluate Model  

```python
# ML libraries
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn import datasets

# Data handling
import pandas as pd

# Preprocessing
from sklearn.preprocessing import StandardScaler

# Evaluation
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Load iris dataset
df = datasets.load_iris(as_frame=True).frame

df.head()
```

```python
# Split features and target

X = df.drop("target", axis=1)
y = df["target"]

X.head()
```

```python
# Split dataset

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

```python
# SVM is sensitive to scaling

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Default SVM uses RBF kernel

svc = SVC()
svc.fit(X_train_scaled, y_train)

y_pred = svc.predict(X_test_scaled)

print("Accuracy (RBF Default):", accuracy_score(y_test, y_pred))
```

```python
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
```

```python
# Linear kernel

svc = SVC(kernel="linear")
svc.fit(X_train_scaled, y_train)

y_pred = svc.predict(X_test_scaled)

print("Accuracy (Linear):", accuracy_score(y_test, y_pred))
```

```python
# Polynomial kernel

svc = SVC(kernel="poly")
svc.fit(X_train_scaled, y_train)

y_pred = svc.predict(X_test_scaled)

print("Accuracy (Polynomial):", accuracy_score(y_test, y_pred))
```

```python
# Sigmoid kernel

svc = SVC(kernel="sigmoid")
svc.fit(X_train_scaled, y_train)

y_pred = svc.predict(X_test_scaled)

print("Accuracy (Sigmoid):", accuracy_score(y_test, y_pred))
```

```python
# Try different values of C

C_vals = [0.5, 1, 2, 3, 4, 5, 6]

for C_val in C_vals:
    svc = SVC(C=C_val, kernel="rbf")
    svc.fit(X_train_scaled, y_train)
    
    acc = svc.score(X_test_scaled, y_test)
    
    print(f"C={C_val}, Accuracy={acc}")
```

## 📊 Observations

- SVM performs very well on small datasets  
- RBF kernel gives strong performance  
- Linear kernel works when data is separable  
- C significantly affects model performance  

## ✅ Conclusion

- SVM is powerful for classification tasks  
- Kernel choice is important  
- Feature scaling is mandatory  
- Hyperparameter tuning improves results
