# 🎯 Choosing Optimal K in K-Means

## 📌 Objective
Find the best number of clusters (K) using:
- Elbow Method
- Silhouette Score

## 🧠 Why Important?
K-Means requires K as input → wrong K = bad clustering

## 🔷 Step 1: Import Libraries

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

## 🔷 Step 2: Create Dataset

We create synthetic data with unknown cluster structure.

```python
# Create dataset
X, y = make_blobs(n_samples=300, centers=4, random_state=42)
```

## 🔷 Step 3: Visualize Data

```python
# Plot raw data
plt.scatter(X[:, 0], X[:, 1])
plt.title("Original Data")
plt.show()
```

## 🔷 Step 4: Elbow Method

We calculate inertia for different values of K.

```python
inertia_values = []

# Try K from 1 to 10
for k in range(1, 11):
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(X)
    inertia_values.append(model.inertia_)
```

## 🔷 Step 5: Plot Elbow Curve

```python
# Plot elbow graph
plt.plot(range(1, 11), inertia_values, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.show()
```

## 🔷 Step 6: Silhouette Score

We compute silhouette score for different K values.

```python
silhouette_scores = []

# Silhouette requires at least 2 clusters
for k in range(2, 11):
    model = KMeans(n_clusters=k, random_state=42)
    labels = model.fit_predict(X)
    score = silhouette_score(X, labels)
    silhouette_scores.append(score)
```

## 🔷 Step 7: Plot Silhouette Scores

```python
# Plot silhouette scores
plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.title("Silhouette Score vs K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.show()
```

## 🔷 Step 8: Final Observation

- Elbow Method → look for bend  
- Silhouette Score → choose highest value  

👉 Best K is where:
- elbow appears  
- silhouette score is high  

Use both together for better decision.
