# 🎯 K-Means Clustering

## 📌 Objective
Group similar data points into clusters using K-Means algorithm.

## 🧠 What You’ll Learn
- How K-Means works practically
- How to apply clustering using sklearn
- How to visualize clusters
- How to evaluate clustering (inertia + elbow method)

## 🔷 Step 1: Import Libraries
We import required libraries for data handling and visualization.

```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 🔷 Step 2: Create Dataset

We generate synthetic data using `make_blobs` for clear clustering visualization.

```python
from sklearn.datasets import make_blobs

# Create dataset with 3 clusters
X, y = make_blobs(n_samples=300, centers=3, random_state=42)
```

## 🔷 Step 3: Visualize Original Data

Before clustering, we visualize the raw data.

```python
# Plot original data (no clustering yet)
plt.scatter(X[:, 0], X[:, 1])
plt.title("Original Data")
plt.show()
```

## 🔷 Step 4: Apply K-Means

We create and train the K-Means model with K=3 clusters.

```python
from sklearn.cluster import KMeans

# Initialize K-Means model
kmeans = KMeans(n_clusters=3, random_state=42)

# Train model
kmeans.fit(X)

# Predict cluster labels
y_pred = kmeans.predict(X)
```

## 🔷 Step 5: Visualize Clusters

Now we visualize how K-Means grouped the data.

```python
# Plot clustered data
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='viridis')
plt.title("K-Means Clustering")
plt.show()
```

## 🔷 Step 6: Plot Centroids

Centroids represent the center of each cluster.

```python
# Get centroids
centroids = kmeans.cluster_centers_

# Plot clusters with centroids
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='viridis')
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', s=200, marker='X')

plt.title("Clusters with Centroids")
plt.show()
```

## 🔷 Step 7: Evaluate Model (Inertia)

Inertia measures how compact the clusters are.

👉 Lower inertia = better clustering

```python
# Print inertia
print("Inertia:", kmeans.inertia_)
```

## 🔷 Step 8: Elbow Method (Choosing K)

We try different values of K and observe inertia.

```python
inertia_values = []

# Try different K values
for k in range(1, 10):
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(X)
    inertia_values.append(model.inertia_)
```

## 🔷 Step 9: Plot Elbow Curve

We plot inertia vs number of clusters to find optimal K.

```python
# Plot elbow graph
plt.plot(range(1, 10), inertia_values, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.show()
```
