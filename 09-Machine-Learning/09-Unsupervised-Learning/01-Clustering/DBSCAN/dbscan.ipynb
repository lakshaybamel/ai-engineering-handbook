# 🌐 DBSCAN Clustering

## 📌 Objective
Apply DBSCAN clustering to identify dense regions and detect outliers.

## 🧠 What You’ll Learn
- How DBSCAN works in practice
- How to form clusters based on density
- How to detect noise points

## 🔷 Step 1: Import Libraries

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
```

## 🔷 Step 2: Create Dataset

We use a non-linear dataset (make_moons)  
👉 This shows advantage of DBSCAN over K-Means

```python
# Create non-linear dataset
X, y = make_moons(n_samples=300, noise=0.05, random_state=42)
```

## 🔷 Step 3: Visualize Data

```python
# Plot original data
plt.scatter(X[:, 0], X[:, 1])
plt.title("Original Data")
plt.show()
```

## 🔷 Step 4: Apply DBSCAN

Parameters:
- eps → neighborhood radius  
- min_samples → minimum points to form cluster

```python
# Apply DBSCAN
model = DBSCAN(eps=0.2, min_samples=5)

# Predict cluster labels
y_pred = model.fit_predict(X)
```

## 🔷 Step 5: Visualize Clusters

👉 DBSCAN assigns:
- Cluster labels (0,1,2...)
- Noise = -1

```python
# Plot clustered data
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='rainbow')
plt.title("DBSCAN Clustering")
plt.show()
```

## 🔷 Step 6: Identify Noise Points

Points labeled as -1 are considered noise (outliers).

```python
# Count noise points
noise_points = np.sum(y_pred == -1)

print("Number of noise points:", noise_points)
```

## 🔷 Final Observations

- DBSCAN detects clusters of non-linear shape  
- It automatically finds number of clusters  
- It identifies noise points  

👉 Useful when data is not spherical
