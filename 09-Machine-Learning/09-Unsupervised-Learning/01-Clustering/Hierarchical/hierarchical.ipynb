# 🌳 Hierarchical Clustering

## 📌 Objective
Apply hierarchical clustering and visualize cluster formation using a dendrogram.

## 🧠 What You’ll Learn
- How hierarchical clustering works
- How to build dendrogram
- How to choose number of clusters

## 🔷 Step 1: Import Libraries

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
```

## 🔷 Step 2: Create Dataset

We generate synthetic data for clustering.

```python
# Create dataset
X, y = make_blobs(n_samples=300, centers=4, random_state=42)
```

## 🔷 Step 3: Visualize Data

```python
# Plot original data
plt.scatter(X[:, 0], X[:, 1])
plt.title("Original Data")
plt.show()
```

## 🔷 Step 4: Build Dendrogram

Dendrogram helps visualize how clusters are merged.

```python
# Create linkage matrix
Z = linkage(X, method='ward')

# Plot dendrogram
plt.figure(figsize=(10, 5))
dendrogram(Z)

plt.title("Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
```

## 🔷 Step 5: Choose Number of Clusters

👉 Draw a horizontal line on dendrogram  
👉 Count number of vertical cuts  

This gives optimal number of clusters.

## 🔷 Step 6: Apply Hierarchical Clustering

We use Agglomerative Clustering with selected number of clusters.

```python
# Apply Agglomerative Clustering
model = AgglomerativeClustering(n_clusters=4)

# Predict cluster labels
y_pred = model.fit_predict(X)
```

## 🔷 Step 7: Visualize Clusters

```python
# Plot clustered data
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='rainbow')
plt.title("Hierarchical Clustering")
plt.show()
```

## 🔷 Final Observations

- Dendrogram helps decide number of clusters  
- Hierarchical clustering does not require K initially  
- Works well for small datasets
