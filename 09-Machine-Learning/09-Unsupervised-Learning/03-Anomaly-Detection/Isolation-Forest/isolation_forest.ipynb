# 🌲 Isolation Forest — Anomaly Detection

## 🎯 Objective
Detect anomalies (outliers) using Isolation Forest.

👉 Points that are rare and different are detected as anomalies.

```python
# Import required libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import IsolationForest
from sklearn.datasets import make_blobs
```

## 📊 Step 1: Create Sample Dataset

We create a dataset with:
- Normal data points
- Some anomalies (outliers)

```python
# Create normal data (clusters)

X, _ = make_blobs(n_samples=300, centers=1, cluster_std=1.0, random_state=42)

# Add outliers manually
outliers = np.random.uniform(low=-6, high=6, size=(20, 2))

# Combine normal + outliers
X = np.vstack([X, outliers])
```

## 📊 Step 2: Visualize Data

```python
plt.scatter(X[:, 0], X[:, 1])
plt.title("Dataset with Outliers")
plt.show()
```

## ⚙️ Step 3: Apply Isolation Forest

```python
# Initialize model

model = IsolationForest(contamination=0.05, random_state=42)

# Train model
model.fit(X)
```

## 🔍 Step 4: Predict Anomalies

Output:
- 1 → Normal point  
- -1 → Anomaly  

```python
# Predict

labels = model.predict(X)

# Convert to DataFrame for clarity
df = pd.DataFrame(X, columns=["Feature1", "Feature2"])
df["Anomaly"] = labels

df.head()
```

## 📊 Step 5: Visualize Anomalies

```python
# Separate normal and anomaly points

normal = df[df["Anomaly"] == 1]
anomaly = df[df["Anomaly"] == -1]

# Plot

plt.scatter(normal["Feature1"], normal["Feature2"], label="Normal")
plt.scatter(anomaly["Feature1"], anomaly["Feature2"], color='red', label="Anomaly")

plt.legend()
plt.title("Isolation Forest Anomaly Detection")
plt.show()
```

## ⚠️ Important Observations

- Anomalies are detected as isolated points  
- They are far from dense regions  
- Model works without labels  

---

## 🎯 Key Takeaways

- Isolation Forest isolates anomalies quickly  
- No distance calculation required  
- Works well for large datasets
