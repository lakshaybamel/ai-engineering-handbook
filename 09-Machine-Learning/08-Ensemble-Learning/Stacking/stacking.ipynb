# 🧱 Stacking Classifier

## 📌 Objective
Combine multiple models using Stacking and improve prediction using a meta-model.

## 📊 Dataset
Wine Dataset (sklearn)

## ⚙️ Workflow
1. Load Dataset  
2. Train Base Models  
3. Apply Stacking  
4. Evaluate Model  

```python
# Base models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier

# Stacking
from sklearn.ensemble import StackingClassifier

# Dataset
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

# Evaluation
from sklearn.metrics import accuracy_score

import pandas as pd
```

```python
# Load dataset

data = load_wine(as_frame=True)
df = data.frame

df.head()
```

```python
# Separate features and target

X = df.drop("target", axis=1)
y = df["target"]
```

```python
# Split dataset

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

## 🤖 Base Models

These models will generate predictions used by the meta-model.

```python
# Define base models

estimators = [
    ('lr', LogisticRegression(max_iter=1000)),
    ('dt', DecisionTreeClassifier(random_state=42)),
    ('knn', KNeighborsClassifier())
]
```

## 🧱 Applying Stacking

Meta-model learns from base model predictions.

```python
# Meta model

meta_model = LogisticRegression()

# Stacking classifier

model = StackingClassifier(
    estimators=estimators,
    final_estimator=meta_model
)

model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)
```

```python
# Evaluate model

print("Stacking Accuracy:", accuracy_score(y_test, y_pred))
```

```python
# Compare with individual models

for name, est in estimators:
    est.fit(X_train, y_train)
    y_pred = est.predict(X_test)
    print(f"{name} Accuracy:", accuracy_score(y_test, y_pred))
```

## 📊 Observations

- Stacking combines multiple models using a meta-model  
- Often performs better than individual models  
- Learns how to weight different models  

## ✅ Conclusion

- Stacking is a powerful ensemble technique  
- Uses meta-learning to improve predictions  
- More flexible than voting
