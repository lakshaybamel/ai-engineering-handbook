# 🧠 Building a Neuron with PyTorch

In this notebook we will build a **single artificial neuron using PyTorch**.

A neuron performs three main steps:

1️⃣ Multiply inputs with weights  
2️⃣ Add a bias  
3️⃣ Produce an output

Mathematically:
```
z = w₁x₁ + w₂x₂ + w₃x₃ + b
```
PyTorch provides a built-in layer called:

nn.Linear()

which represents a **fully connected neuron**.

## ⚙️ Step 1: Import Libraries

We first import **PyTorch** and its neural network module.

````python
import torch
import torch.nn as nn
````

## 📥 Step 2: Create Input Tensor

We create a tensor representing **three input features**.

Example inputs:
- x₁
- x₂
- x₃

````python
# Set seed for reproducibility
torch.manual_seed(42)

# Input tensor
inputs = torch.tensor([1.0, 2.0, 3.0])
````

## 🧠 Step 3: Create a Neuron

We create a neuron using:

nn.Linear(in_features, out_features)

Parameters:

in_features → number of inputs  
out_features → number of neurons

Since we are building **one neuron**, the output size is 1.

````python
# Create a single neuron
neuron = nn.Linear(in_features=3, out_features=1)
````

## 🚀 Step 4: Forward Pass

Now we pass the inputs through the neuron.

This performs:
```
z = w₁x₁ + w₂x₂ + w₃x₃ + b
```

````python
output = neuron(inputs)

print("Neuron Output:", output)
````

## 🔍 Step 5: Verify the Formula Manually

We verify that PyTorch computes the same result using:
```
z = w · x + b
```

````python
# Manual computation
neuron.weight @ inputs + neuron.bias
````

## ⚙️ Step 6: Inspect Weights

````python
neuron.weight
````

## ⚙️ Step 7: Inspect Bias

````python
neuron.bias
````

## 📊 Summary

In this notebook we learned:

✔ A neuron performs a weighted sum of inputs  
✔ PyTorch provides `nn.Linear()` to represent a neuron  
✔ The neuron internally calculates:

z = w₁x₁ + w₂x₂ + w₃x₃ + b

This simple neuron is the **basic building block of neural networks**.

Multiple neurons combined together form **deep neural networks**.
