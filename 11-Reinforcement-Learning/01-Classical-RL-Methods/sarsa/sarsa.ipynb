# 🔁 SARSA Algorithm Implementation

In this notebook, we implement the **SARSA (State-Action-Reward-State-Action)** algorithm.

SARSA is an **on-policy reinforcement learning algorithm** that learns the value of actions based on the **actual actions taken by the agent**.

We will apply SARSA to a simple **Grid World / Cliff Walking-like environment**.

---

## 🎯 Objective

- Learn optimal Q-values  
- Build a safe policy  
- Understand on-policy learning  

## 🧠 Key Idea of SARSA

SARSA follows this sequence:
```
State → Action → Reward → Next State → Next Action
```
Update rule:
```math
Q(s, a) = Q(s, a) + α [ R + γQ(s', a') - Q(s, a) ]
```
👉 Uses **actual next action (a')**

## Step 1: Import Libraries

````python
import numpy as np
import random
````

## Step 2: Define Environment

We create a simple **Grid World**:

- Start state
- Goal state
- Cliff (danger zone)

````python
# Grid size
rows = 4
cols = 6

# Define actions
actions = ['up', 'down', 'left', 'right']

# Action mapping
action_map = {
    'up': (-1, 0),
    'down': (1, 0),
    'left': (0, -1),
    'right': (0, 1)
}

# Start and Goal
start = (3, 0)
goal = (3, 5)

# Cliff positions
cliff = [(3, i) for i in range(1, 5)]
````

## Step 3: Initialize Q-Table

````python
# Q-table: state → action values
Q = {}

for i in range(rows):
    for j in range(cols):
        Q[(i, j)] = {a: 0.0 for a in actions}
````

## Step 4: Hyperparameters

````python
alpha = 0.1      # learning rate
gamma = 0.9      # discount factor
epsilon = 0.1    # exploration rate
episodes = 500
````

## Step 5: ε-Greedy Policy

- Explore with probability ε  
- Exploit best action otherwise  

````python
def choose_action(state):
    if random.uniform(0, 1) < epsilon:
        return random.choice(actions)  # explore
    else:
        return max(Q[state], key=Q[state].get)  # exploit
````

## Step 6: Environment Step Function

````python
def step(state, action):
    i, j = state
    di, dj = action_map[action]

    new_state = (i + di, j + dj)

    # Boundary check
    if new_state[0] < 0 or new_state[0] >= rows or new_state[1] < 0 or new_state[1] >= cols:
        return state, -1

    # Cliff check
    if new_state in cliff:
        return start, -100

    # Goal check
    if new_state == goal:
        return new_state, 10

    return new_state, -1
````

## Step 7: SARSA Training Loop

````python
for ep in range(episodes):

    state = start
    action = choose_action(state)

    while state != goal:

        next_state, reward = step(state, action)
        next_action = choose_action(next_state)

        # SARSA Update Rule
        Q[state][action] = Q[state][action] + alpha * (
            reward + gamma * Q[next_state][next_action] - Q[state][action]
        )

        state = next_state
        action = next_action

    if (ep+1) % 100 == 0:
        print(f"Episode {ep+1} completed")
````

## Step 8: Extract Policy

````python
policy = {}

for state in Q:
    policy[state] = max(Q[state], key=Q[state].get)

policy
````

## Step 9: Visualize Policy

````python
grid_policy = []

for i in range(rows):
    row = []
    for j in range(cols):
        if (i, j) == goal:
            row.append('G')
        elif (i, j) in cliff:
            row.append('X')
        else:
            row.append(policy[(i, j)][0].upper())
    grid_policy.append(row)

for r in grid_policy:
    print(r)
````

## 🎯 Observations

- SARSA learns a **safe path**
- It avoids the cliff region
- Even if shorter path exists, it prefers safer route

---

## 🧠 Why?

Because SARSA is **on-policy**:

- It considers exploration during learning  
- Risky states get penalized
