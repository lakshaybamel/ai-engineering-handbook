# ⚡ Q-Learning Algorithm Implementation

In this notebook, we implement the **Q-Learning algorithm**, one of the most important reinforcement learning techniques.

Q-learning is an **off-policy algorithm** that learns the optimal policy by always considering the **best possible future action**.

---

## 🎯 Objective

- Learn optimal Q-values  
- Compare behavior with SARSA  
- Understand off-policy learning  
- Observe optimal (possibly risky) policy  

## 🧠 Key Idea of Q-Learning

Q-learning update rule:

```math
Q(s, a) = Q(s, a) + α [ R + γ max Q(s', a') - Q(s, a) ]
```

👉 Uses **maximum future reward** instead of actual next action

This makes Q-learning:

- Off-policy  
- Greedy  
- Optimal (but sometimes risky)  

## Step 1: Import Libraries

````python
import numpy as np
import random
````

## Step 2: Define Environment

We create a simple Grid World environment:

- Start state  
- Goal state  
- Cliff (danger zone)  

````python
# Grid size
rows = 4
cols = 6

# Actions
actions = ['up', 'down', 'left', 'right']

# Action mapping
action_map = {
    'up': (-1, 0),
    'down': (1, 0),
    'left': (0, -1),
    'right': (0, 1)
}

# Start and Goal
start = (3, 0)
goal = (3, 5)

# Cliff positions
cliff = [(3, i) for i in range(1, 5)]
````

## Step 3: Initialize Q-Table

````python
Q = {}

for i in range(rows):
    for j in range(cols):
        Q[(i, j)] = {a: 0.0 for a in actions}
````

## Step 4: Hyperparameters

````python
alpha = 0.1      # learning rate
gamma = 0.9      # discount factor
epsilon = 0.1    # exploration rate
episodes = 500
````

## Step 5: ε-Greedy Policy

- Explore with probability ε  
- Exploit best action otherwise  

````python
def choose_action(state):
    if random.uniform(0, 1) < epsilon:
        return random.choice(actions)
    else:
        return max(Q[state], key=Q[state].get)
````

## Step 6: Environment Step Function

````python
def step(state, action):
    i, j = state
    di, dj = action_map[action]

    new_state = (i + di, j + dj)

    # Boundary check
    if new_state[0] < 0 or new_state[0] >= rows or new_state[1] < 0 or new_state[1] >= cols:
        return state, -1

    # Cliff check
    if new_state in cliff:
        return start, -100

    # Goal check
    if new_state == goal:
        return new_state, 10

    return new_state, -1
````

## Step 7: Q-Learning Training Loop

````python
for ep in range(episodes):

    state = start

    while state != goal:

        action = choose_action(state)
        next_state, reward = step(state, action)

        # Q-Learning Update Rule
        best_next_action = max(Q[next_state], key=Q[next_state].get)

        Q[state][action] = Q[state][action] + alpha * (
            reward + gamma * Q[next_state][best_next_action] - Q[state][action]
        )

        state = next_state

    if (ep+1) % 100 == 0:
        print(f"Episode {ep+1} completed")
````

## Step 8: Extract Learned Policy

````python
policy = {}

for state in Q:
    policy[state] = max(Q[state], key=Q[state].get)

policy
````

## Step 9: Visualize Policy

````python
grid_policy = []

for i in range(rows):
    row = []
    for j in range(cols):
        if (i, j) == goal:
            row.append('G')
        elif (i, j) in cliff:
            row.append('X')
        else:
            row.append(policy[(i, j)][0].upper())
    grid_policy.append(row)

for r in grid_policy:
    print(r)
````

## 🎯 Observations

- Q-learning learns the **shortest path**
- It prefers paths near the cliff
- It ignores exploration risk during updates

---

## ⚠️ Why Risky?

Because Q-learning:

- assumes optimal future actions  
- ignores random exploration  

## 🔥 SARSA vs Q-Learning Insight

| Feature | SARSA | Q-Learning |
|--------|------|-----------|
| Type | On-policy | Off-policy |
| Behavior | Safe | Optimal |
| Risk | Low | High |
| Update uses | Actual action | Best action |

---

👉 Q-learning is more **aggressive and goal-oriented**
