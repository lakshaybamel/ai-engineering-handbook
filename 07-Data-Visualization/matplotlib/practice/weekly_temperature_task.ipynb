# Practice Task — Weekly Temperature Analysis

Goal:
Analyze temperature trends of a city over a week using different plot types.

We will:

- visualize trend
- compare weekdays vs weekend
- analyze distribution
- identify peak temperature

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [30, 32, 35, 33, 31, 28, 29]
```

```python
plt.figure(figsize=(8,4))

plt.plot(days, temperature, marker='o')
plt.title("Weekly Temperature Trend")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.grid(True)

plt.show()
```

```python
max_temp = max(temperature)
max_day = days[temperature.index(max_temp)]

plt.plot(days, temperature, marker='o')
plt.scatter(max_day, max_temp, color="red", s=120)

plt.annotate("Highest",
             xy=(max_day, max_temp),
             xytext=(4, max_temp+2),
             arrowprops=dict(arrowstyle="->"))

plt.title("Peak Temperature Highlighted")
plt.show()
```

```python
plt.bar(days, temperature, color="skyblue")

plt.title("Temperature Comparison")
plt.ylabel("°C")

plt.show()
```

```python
weekday_avg = np.mean(temperature[:5])
weekend_avg = np.mean(temperature[5:])

labels = ["Weekday Avg", "Weekend Avg"]
values = [weekday_avg, weekend_avg]

plt.bar(labels, values, color=["blue", "green"])

plt.title("Weekday vs Weekend Temperature")
plt.ylabel("Average °C")

plt.show()
```

```python
plt.hist(temperature, bins=5, color="orange", edgecolor="black")

plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.boxplot(temperature, showmeans=True)

plt.title("Temperature Spread")
plt.show()
```

From this analysis:

- Highest temperature occurred mid-week
- Weekend temperatures were lower
- Distribution is slightly right-skewed
- No extreme outliers

This demonstrates how multiple visualizations provide deeper insight.
