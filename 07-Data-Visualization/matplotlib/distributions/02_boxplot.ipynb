# Box Plot

Box plots summarize distribution using quartiles.

They show:

- median
- quartiles (Q1, Q3)
- range
- outliers

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(1)

data = np.random.normal(50, 10, 100)
```

```python
plt.boxplot(data)

plt.title("Basic Box Plot")
plt.show()
```

```python
plt.boxplot(data, vert=False)

plt.title("Horizontal Box Plot")
plt.show()
```

```python
plt.boxplot(data, showmeans=True)

plt.title("Box Plot with Mean")
plt.show()
```

```python
# add artificial outliers
data_with_outliers = np.append(data, [100, 105, 110])

plt.boxplot(data_with_outliers)

plt.title("Outliers Visible")
plt.show()
```

```python
data2 = np.random.normal(60, 8, 100)

plt.boxplot([data, data2])

plt.xticks([1,2], ["Group A", "Group B"])
plt.title("Multiple Groups Comparison")

plt.show()
```

```python
plt.boxplot(data,
            patch_artist=True,
            boxprops=dict(facecolor="lightblue"),
            medianprops=dict(color="red"))

plt.title("Styled Box Plot")
plt.show()
```

Box plots are powerful for:

- detecting outliers
- comparing distributions
- checking spread

Widely used in data preprocessing stage.
