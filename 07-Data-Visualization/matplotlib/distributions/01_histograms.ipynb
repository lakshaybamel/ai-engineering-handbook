# Histograms

A histogram shows distribution of numerical data.

It divides data into bins and counts frequency.

Used for:

- checking distribution
- identifying skewness
- detecting outliers

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(0)

data = np.random.normal(loc=50, scale=10, size=1000)
```

```python
plt.hist(data)

plt.title("Basic Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.hist(data, bins=20, color="skyblue", edgecolor="black")

plt.title("Styled Histogram")
plt.show()
```

```python
data2 = np.random.normal(60, 8, 1000)

plt.hist(data, bins=20, alpha=0.6, label="Dataset 1")
plt.hist(data2, bins=20, alpha=0.6, label="Dataset 2")

plt.legend()
plt.title("Multiple Histograms")
plt.show()
```

```python
mean_value = np.mean(data)

plt.hist(data, bins=20, color="lightgreen", edgecolor="black")

plt.axvline(mean_value, color="red", linestyle="--", linewidth=2)

plt.title("Histogram with Mean Line")
plt.show()
```

```python
plt.hist(data, bins=20, density=True, alpha=0.7)

plt.title("Density Histogram")
plt.show()
```

Histograms help analyze:

- distribution shape
- central tendency
- spread
- skewness
