# Pie Chart

Pie charts represent proportion of categories as slices of a circle.

Used for:

- percentage distribution
- market share
- population breakdown

```python
import matplotlib.pyplot as plt
```

```python
categories = ["Laptop", "Phone", "Tablet", "Accessories"]
sales = [40, 35, 15, 10]
```

```python
plt.pie(sales)

plt.title("Basic Pie Chart")
plt.show()
```

```python
plt.pie(sales, labels=categories)

plt.title("Pie Chart with Labels")
plt.show()
```

```python
plt.pie(sales, labels=categories, autopct="%1.1f%%")

plt.title("Pie Chart with Percentages")
plt.show()
```

```python
explode = [0.1, 0, 0, 0]  # explode first slice

plt.pie(sales, labels=categories, autopct="%1.1f%%", explode=explode)

plt.title("Exploded Slice")
plt.show()
```

```python
colors = ["gold", "lightblue", "lightgreen", "pink"]

plt.pie(sales,
        labels=categories,
        autopct="%1.1f%%",
        colors=colors)

plt.title("Colored Pie Chart")
plt.show()
```

```python
plt.pie(sales,
        labels=categories,
        autopct="%1.1f%%",
        startangle=90,
        shadow=True)

plt.title("Styled Pie Chart")
plt.show()
```

```python
# create donut effect
plt.pie(sales,
        labels=categories,
        autopct="%1.1f%%")

# draw circle in center
centre_circle = plt.Circle((0,0),0.6,fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title("Donut Chart")
plt.show()
```

Pie charts are best for:

- percentage distribution<br>
- small number of categories<br>

Avoid using pie charts for too many categories.
