# Multiple Line Plots

Multiple lines are used to compare trends.

Example uses:
- compare cities temperature
- compare company revenue
- compare model accuracy

```python
import matplotlib.pyplot as plt
```

```python
days = [1,2,3,4,5,6,7] # Dataset (Two Cities Temperature)

city_A = [30,32,35,33,31,29,28]
city_B = [25,27,30,29,28,26,24]
```

```python
plt.plot(days, city_A)
plt.plot(days, city_B)

plt.title("Temperature Comparison")
plt.show()
```

```python
plt.plot(days, city_A, label="City A")
plt.plot(days, city_B, label="City B")

plt.legend()
plt.title("City Temperature Comparison")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")

plt.show()
```

```python
plt.plot(days, city_A, 'r-o', label="City A")
plt.plot(days, city_B, 'b--s', label="City B")

plt.legend()
plt.grid(True)

plt.title("Styled Comparison")
plt.show()
```

```python
city_C = [20,22,25,24,23,21,19]

plt.plot(days, city_A, label="City A")
plt.plot(days, city_B, label="City B")
plt.plot(days, city_C, label="City C")

plt.legend()
plt.title("Three City Comparison")

plt.show()
```

```python
# Real Case: Company Revenue

years = [2019,2020,2021,2022,2023]

company_X = [10,15,25,40,60]
company_Y = [8,12,18,28,45]

plt.plot(years, company_X, marker='o', label="Company X")
plt.plot(years, company_Y, marker='s', label="Company Y")

plt.legend()
plt.xlabel("Year")
plt.ylabel("Revenue (Million $)")
plt.title("Company Revenue Growth")

plt.show()
```

Multiple line plots help compare behavior.

Key idea:<br>
One graph → multiple insights
