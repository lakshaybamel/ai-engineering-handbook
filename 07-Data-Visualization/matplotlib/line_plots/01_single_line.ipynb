# Single Line Plot

Line plots are used to show change over time.

Very common in:
- stock prices
- temperature
- growth trends

```python
import matplotlib.pyplot as plt
```

```python
days = [1,2,3,4,5,6,7]
temperature = [30, 32, 35, 33, 31, 29, 28]
```

```python
plt.plot(days, temperature)

plt.title("Temperature Over a Week")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")

plt.show()
```

```python
plt.plot(days, temperature, 'g--o')

plt.grid(True)
plt.title("Temperature with Grid")

plt.show()
```

```python
max_temp = max(temperature)
max_day = days[temperature.index(max_temp)]

plt.plot(days, temperature, 'b-o')

plt.scatter(max_day, max_temp, color="red", s=100)

plt.title("Highest Temperature Highlighted")
plt.show()
```

```python
plt.plot(days, temperature, 'b-o')

plt.annotate("Peak",
             xy=(max_day, max_temp),
             xytext=(max_day+0.3, max_temp+1),
             arrowprops=dict(arrowstyle="->"))

plt.title("Annotated Peak")
plt.show()
```

```python
years = [2018,2019,2020,2021,2022,2023]
users = [200,350,500,800,1200,1800]

plt.plot(years, users, marker='o')

plt.title("App User Growth")
plt.xlabel("Year")
plt.ylabel("Users")

plt.show()
```

Line plots help understand trends and growth patterns.

Key ideas:
- ordered data
- change over time
- detecting peaks
