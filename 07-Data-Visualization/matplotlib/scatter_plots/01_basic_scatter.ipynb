# Basic Scatter Plot

Scatter plots are used to show relationship between two variables.

Used for:
- correlation analysis
- detecting patterns
- spotting outliers

```python
import matplotlib.pyplot as plt
```

```python
# Example: Study hours vs Exam score
study_hours = [1,2,3,4,5,6,7,8]
exam_score = [35,40,50,55,60,65,75,85]
```

```python
plt.scatter(study_hours, exam_score)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.show()
```

```python
# Example: Height vs Weight
height = [150,155,160,165,170,175,180,185]
weight = [50,55,60,65,72,78,85,90]

plt.scatter(height, weight)

plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")

plt.show()
```

```python
import numpy as np

np.random.seed(0)

x = np.random.randint(1, 50, 30)
y = x + np.random.randint(-10, 10, 30)

plt.scatter(x, y)

plt.title("Random Scatter Data")
plt.xlabel("X values")
plt.ylabel("Y values")

plt.show()
```

Scatter plots help visualize:

- positive correlation
- negative correlation
- no correlation
- outliers
