# Scatter Plot — Annotations & Multiple Datasets

We often need to:<br>
• highlight specific observations<br>
• compare groups<br>
• label outliers<br>

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Example: Study hours vs marks for two classes
hours_A = [1,2,3,4,5,6,7]
marks_A = [35,40,50,55,60,70,80]

hours_B = [1,2,3,4,5,6,7]
marks_B = [30,38,45,52,58,65,72]
```

```python
plt.scatter(hours_A, marks_A, label="Class A")
plt.scatter(hours_B, marks_B, label="Class B")

plt.legend()
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Class Comparison")

plt.show()
```

```python
plt.scatter(hours_A, marks_A, color="blue", label="Class A")
plt.scatter(hours_B, marks_B, color="green", label="Class B")

plt.legend()
plt.title("Colored Classes")
plt.show()
```

```python
max_mark = max(marks_A)
index = marks_A.index(max_mark)

plt.scatter(hours_A, marks_A)
plt.scatter(hours_A[index], max_mark, color="red", s=120)

plt.title("Top Performer Highlighted")
plt.show()
```

```python
plt.scatter(hours_A, marks_A)

plt.annotate("Top Student",
             xy=(hours_A[index], max_mark),
             xytext=(hours_A[index]+0.5, max_mark+5),
             arrowprops=dict(arrowstyle="->"))

plt.title("Annotated Scatter Plot")
plt.show()
```

```python
np.random.seed(2)

x = np.random.normal(50, 10, 40)
y = x + np.random.normal(0, 8, 40)

plt.scatter(x, y)

# add artificial outlier
plt.scatter(90, 20, color="red", s=150)

plt.annotate("Outlier",
             xy=(90,20),
             xytext=(75,35),
             arrowprops=dict(arrowstyle="->"))

plt.title("Outlier Detection Example")
plt.show()
```

Annotations help explain data.

Important for:<br>
• presentations<br>
• reports<br>
• identifying anomalies<br>
