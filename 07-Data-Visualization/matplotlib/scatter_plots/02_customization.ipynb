# Scatter Plot Customization

We can control:
- color
- size
- transparency
- colormap

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(1)

x = np.random.randint(10, 100, 40)
y = x + np.random.randint(-20, 20, 40)
```

```python
plt.scatter(x, y, color="red")

plt.title("Red Scatter Points")
plt.show()
```

```python
sizes = np.random.randint(20, 200, 40)

plt.scatter(x, y, s=sizes)

plt.title("Variable Size Points")
plt.show()
```

```python
plt.scatter(x, y, alpha=0.5)

plt.title("Transparent Points")
plt.show()
```

```python
colors = np.random.randint(1, 10, 40)

plt.scatter(x, y, c=colors, cmap="viridis")

plt.title("Colormap Visualization")
plt.colorbar()

plt.show()
```

```python
plt.scatter(x, y, color="skyblue", edgecolors="black", linewidths=1)

plt.title("Points with Borders")
plt.show()
```

```python
sizes = np.random.randint(50, 300, 40)
colors = np.random.randint(1, 10, 40)

plt.scatter(x, y, s=sizes, c=colors, cmap="plasma", alpha=0.7, edgecolors="black")

plt.colorbar()
plt.title("Fully Customized Scatter")

plt.show()
```

Scatter customization improves readability and insight.

Color → category or value<br>
Size → magnitude<br>
Transparency → density
