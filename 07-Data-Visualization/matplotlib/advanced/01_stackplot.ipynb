# Stack Plot

Stack plots show cumulative values over time.

Used for:

- revenue breakdown
- resource usage
- category contribution over time

```python
import matplotlib.pyplot as plt
```

```python
years = [2019, 2020, 2021, 2022, 2023]

product_A = [10, 15, 20, 25, 30]
product_B = [5, 10, 15, 20, 25]
product_C = [8, 12, 18, 22, 28]
```

```python
plt.stackplot(years, product_A, product_B, product_C)

plt.title("Revenue Contribution by Product")
plt.xlabel("Year")
plt.ylabel("Revenue")

plt.show()
```

```python
plt.stackplot(years,
              product_A,
              product_B,
              product_C,
              labels=["Product A", "Product B", "Product C"])

plt.legend(loc="upper left")
plt.title("Stack Plot with Legend")

plt.show()
```

```python
colors = ["skyblue", "lightgreen", "salmon"]

plt.stackplot(years,
              product_A,
              product_B,
              product_C,
              labels=["Product A", "Product B", "Product C"],
              colors=colors)

plt.legend(loc="upper left")
plt.title("Colored Stack Plot")

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]

solar = [5, 8, 12, 15, 20]
wind = [10, 12, 14, 18, 22]
hydro = [20, 25, 30, 35, 40]

plt.stackplot(months, solar, wind, hydro,
              labels=["Solar", "Wind", "Hydro"])

plt.legend()
plt.title("Energy Production Sources")

plt.show()
```

Stack plots show how individual components contribute to a total over time.

Helpful for:
    
- trend analysis
- cumulative growth understanding
