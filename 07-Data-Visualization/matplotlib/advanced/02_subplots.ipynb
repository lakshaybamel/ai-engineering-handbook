# Subplots in Matplotlib

Subplots allow multiple plots inside one figure.

Used for:

- dashboards
- comparisons
- reporting

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(1, 6)
y1 = x * 2
y2 = x ** 2
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)  # 1 row, 2 columns, first plot
plt.plot(x, y1)
plt.title("Linear")

plt.subplot(1,2,2)  # second plot
plt.plot(x, y2)
plt.title("Quadratic")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.subplot(2,2,1)
plt.plot(x, y1)
plt.title("Plot 1")

plt.subplot(2,2,2)
plt.bar(x, y1)
plt.title("Plot 2")

plt.subplot(2,2,3)
plt.scatter(x, y2)
plt.title("Plot 3")

plt.subplot(2,2,4)
plt.hist(y2)
plt.title("Plot 4")

plt.tight_layout()
plt.show()
```

```python
# Using plt.subplots() (Modern Way)

fig, ax = plt.subplots(1, 2, figsize=(10,4))

ax[0].plot(x, y1)
ax[0].set_title("Linear")

ax[1].plot(x, y2)
ax[1].set_title("Quadratic")

plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(2, 2, figsize=(8,6))

axs[0,0].plot(x, y1)
axs[0,0].set_title("Line")

axs[0,1].bar(x, y1)
axs[0,1].set_title("Bar")

axs[1,0].scatter(x, y2)
axs[1,0].set_title("Scatter")

axs[1,1].hist(y2)
axs[1,1].set_title("Histogram")

plt.tight_layout()
plt.show()
```

Subplots allow multiple visualizations in one figure.

Preferred method:<br>
fig, ax = plt.subplots()

It provides more control and cleaner code.
