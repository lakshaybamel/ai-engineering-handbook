# Modern Matplotlib (Object Oriented Style)

Old way:<br>
plt.plot()

Modern way:<br>
fig, ax = plt.subplots()

Modern matplotlib is:

- cleaner
- scalable
- suitable for dashboards

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(1, 11)
y = x * 2
```

```python
plt.plot(x, y)
plt.title("Old Style Plot")
plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(x, y)
ax.set_title("Modern OO Plot")
ax.set_xlabel("X")
ax.set_ylabel("Y")

plt.show()
```

```python
# Multiple Plots Using OO

y2 = x ** 2

fig, ax = plt.subplots()

ax.plot(x, y, label="Linear")
ax.plot(x, y2, label="Quadratic")

ax.legend()
ax.set_title("Multiple Lines - Modern Way")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,5))

ax.plot(x, y, color="blue", linestyle="--", marker="o")
ax.grid(True)

ax.set_title("Styled Modern Plot")

plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(10,4))

axs[0].plot(x, y)
axs[0].set_title("Linear")

axs[1].plot(x, y2)
axs[1].set_title("Quadratic")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(x, y)
ax.set_title("Saving Figure")

fig.savefig("modern_plot.png", dpi=300)
plt.show()
```

Modern matplotlib uses:

fig → figure<br>
ax → axes<br>

Always prefer object oriented style in real projects.
