# Vertical Bar Charts

Bar charts compare values between categories.

Used when:
- comparing products
- comparing students marks
- comparing sales

```python
import matplotlib.pyplot as plt
```

```python
subjects = ["Math", "Physics", "Chemistry", "English", "CS"]
marks = [85, 78, 92, 74, 95]
```

```python
plt.bar(subjects, marks)

plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```

```python
colors = ["red", "blue", "green", "orange", "purple"]

plt.bar(subjects, marks, color=colors)

plt.title("Colored Bar Chart")
plt.show()
```

```python
plt.bar(subjects, marks, width=0.5)

plt.title("Narrow Bars")
plt.show()
```

```python
plt.bar(subjects, marks, color="skyblue", edgecolor="black")

plt.title("Bar Borders")
plt.show()
```

```python
products = ["Laptop", "Phone", "Tablet", "Headphones"]
sales = [120, 300, 150, 200]

plt.bar(products, sales, color="teal")

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Units Sold")

plt.show()
```

```python
plt.bar(products, sales)

plt.xticks(rotation=45)
plt.title("Rotated Labels")

plt.show()
```

Bar charts are best for categorical comparison.

Key idea:<br>
Height = Value<br>
Category = Label<br>
