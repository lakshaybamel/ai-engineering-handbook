# Horizontal Bar Charts

Horizontal bar charts are preferred when category names are long.<br>
They improve readability compared to vertical bars

```python
import matplotlib.pyplot as plt
```

```python
# Example: programming language popularity
languages = ["Python", "Java", "C++", "JavaScript", "Go"]
users = [95, 80, 70, 85, 60]
```

```python
# barh() creates horizontal bars
plt.barh(languages, users)

plt.title("Programming Language Popularity")
plt.xlabel("Users (%)")
plt.ylabel("Language")

plt.show()
```

```python
# different color for each bar
colors = ["green", "blue", "orange", "purple", "gray"]

plt.barh(languages, users, color=colors)

plt.title("Colored Horizontal Bars")
plt.show()
```

```python
plt.barh(languages, users, color="skyblue")

# invert so highest value appears at top
plt.gca().invert_yaxis()

plt.title("Top Language First")
plt.show()
```

```python
plt.barh(languages, users, height=0.5)

plt.title("Thinner Bars")
plt.show()
```

```python
countries = ["India", "USA", "China", "Brazil", "UK"]
population = [1420, 331, 1410, 213, 67]  # in millions

plt.barh(countries, population, color="teal")

plt.xlabel("Population (millions)")
plt.title("Country Population Comparison")

plt.show()
```

Use horizontal bars when labels are long.

It improves readability and presentation quality.
