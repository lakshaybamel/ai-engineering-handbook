# Multiple Bar Charts & Bar Labels

Used for comparing multiple categories across groups.

Example:
- Sales in different years
- Marks of multiple students

```python
import matplotlib.pyplot as plt
import numpy as np  # for positioning bars
```

```python
# Example: Sales comparison
products = ["Laptop", "Phone", "Tablet"]

sales_2022 = [100, 200, 150]
sales_2023 = [120, 250, 180]

# Create index positions
x = np.arange(len(products))

# Width of each bar
width = 0.35
```

```python
# Create figure
plt.figure(figsize=(8,5))

# First dataset bars
plt.bar(x - width/2, sales_2022, width, label="2022")

# Second dataset bars
plt.bar(x + width/2, sales_2023, width, label="2023")

# Labels and title
plt.xticks(x, products)
plt.ylabel("Sales")
plt.title("Sales Comparison by Year")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8,5))

bars1 = plt.bar(x - width/2, sales_2022, width, label="2022")
bars2 = plt.bar(x + width/2, sales_2023, width, label="2023")

plt.xticks(x, products)
plt.legend()

# Add labels above bars
for bar in bars1:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height,
             str(height),
             ha='center',
             va='bottom')

for bar in bars2:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height,
             str(height),
             ha='center',
             va='bottom')

plt.title("Grouped Bar Chart with Labels")

plt.show()
```

```python
students = ["Alice", "Bob", "Charlie"]

math = [85, 78, 90]
science = [88, 82, 91]

x = np.arange(len(students))
width = 0.3

plt.figure(figsize=(8,5))

plt.bar(x - width/2, math, width, label="Math")
plt.bar(x + width/2, science, width, label="Science")

plt.xticks(x, students)
plt.ylabel("Marks")
plt.title("Student Performance Comparison")
plt.legend()

plt.show()
```

Grouped bar charts help compare:
- Multiple categories
- Multiple years
- Multiple groups

Adding value labels improves readability in reports.
