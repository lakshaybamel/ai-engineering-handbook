# Format Strings in Matplotlib

Matplotlib allows styling plots using format strings.

Format = color + marker + line style

Example:
'r--o'
r = red
-- = dashed line
o = circle marker

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]
```

```python
plt.plot(x, y, 'r')   # red
plt.title("Red Line")
plt.show()
```

```python
plt.plot(x, y, '--')  # dashed
plt.title("Dashed Line")
plt.show()
```

```python
plt.plot(x, y, 'o')  # circle markers
plt.title("Markers Only")
plt.show()
```

```python
plt.plot(x, y, 'g--o')  # green dashed with circle
plt.title("Combined Format")
plt.show()
```

Format order:<br>
color → line → marker

```python
plt.plot(x, y, 'b-o', label="Blue")
plt.plot(x, [3,5,7,9,11], 'r--s', label="Red Square")
plt.plot(x, [1,2,3,4,5], 'k:^', label="Black Triangle")

plt.legend()
plt.title("Multiple Styles")
plt.show()
```

```python
plt.plot(x, y, 'g-o', linewidth=3, markersize=10)

plt.title("Custom Thickness & Size")
plt.show()
```

```python
plt.plot(x, y, color="purple", marker="*", linestyle=":")
plt.title("Named Color & Star Marker")
plt.show()
```

```python
plt.plot(x, y, color="#FF5733", marker="D")
plt.title("Hex Color Code")
plt.show()
```
