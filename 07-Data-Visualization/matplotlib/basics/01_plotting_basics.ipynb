# Plotting Basics using Matplotlib

This notebook introduces the fundamentals of plotting in Python.

We will learn:
- How to create a plot
- Line plots
- Multiple datasets
- Labels and titles

```python
import matplotlib.pyplot as plt
```

```python
# simple data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.show()
```

Matplotlib connects points in order and forms a line graph.

```python
plt.plot(x, y)

plt.title("Simple Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.show()
```

```python
y2 = [1, 3, 5, 7, 9]

plt.plot(x, y, label="Even Pattern")
plt.plot(x, y2, label="Odd Pattern")

plt.title("Multiple Lines")
plt.xlabel("Numbers")
plt.ylabel("Values")
plt.legend()

plt.show()
```

```python
plt.plot(x, y, 'r--', label="Red Dashed")
plt.plot(x, y2, 'g-o', label="Green Circle")

plt.legend()
plt.title("Formatted Lines")

plt.show()
```

```python
plt.plot(x, y)
plt.title("Saved Plot")

plt.savefig("my_plot.png")
plt.show()
```

We learned:

- How to create plots
- Add labels & legends
- Multiple datasets
- Styling
- Save figures
