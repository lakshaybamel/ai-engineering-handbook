# Styling & Saving Plots

In real analysis, plots must be readable and presentable.

We will learn:
- figure size
- grid
- background styles
- ticks & limits
- saving plots


```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,20,25,30,40]
```

```python
plt.figure(figsize=(8,5))

plt.plot(x, y)
plt.title("Custom Figure Size")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(x, y)
plt.grid(True)

plt.title("Grid Enabled")
plt.show()
```

```python
plt.style.use("ggplot")

plt.plot(x, y)
plt.title("GGPlot Style")

plt.show()
```

```python
plt.style.use("dark_background")

plt.plot(x, y)
plt.title("Dark Theme")

plt.show()
```

```python
plt.plot(x, y)

plt.xlim(0,6)
plt.ylim(0,50)

plt.title("Custom Axis Limits")
plt.show()
```

```python
plt.plot(x, y)

plt.xticks([1,2,3,4,5])
plt.yticks([0,10,20,30,40,50])

plt.title("Custom Ticks")
plt.show()
```

```python
plt.figure(figsize=(5,3))
plt.plot(x, y)

plt.title("Tight Layout Example")
plt.tight_layout()

plt.show()
```

```python
plt.plot(x, y)
plt.title("Saved Image")

plt.savefig("styled_plot.png", dpi=300)
plt.show()
```

Now you can:

• control figure size<br>
• apply themes<br>
• add grids<br>
• set limits & ticks<br>
• save high-quality images<br>
