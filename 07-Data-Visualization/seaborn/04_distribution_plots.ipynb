# Distribution Plots in Seaborn

Distribution plots help understand how values are spread.

Used for:
- checking normal distribution
- skewness detection
- preprocessing decisions

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
```

```python
sns.histplot(tips["total_bill"])

plt.title("Histogram of Total Bill")
plt.show()
```

```python
sns.histplot(tips["total_bill"], kde=True)

plt.title("Histogram + Density Curve")
plt.show()
```

```python
sns.kdeplot(tips["total_bill"], fill=True)

plt.title("Kernel Density Estimate")
plt.show()
```

```python
sns.histplot(data=tips, x="total_bill", hue="sex", kde=True)

plt.title("Distribution by Gender")
plt.show()
```

```python
sns.kdeplot(x="total_bill", y="tip", data=tips, fill=True)

plt.title("2D Density Distribution")
plt.show()
```

```python
sns.jointplot(x="total_bill", y="tip", data=tips, kind="scatter")

plt.show()
```

```python
sns.pairplot(tips)

plt.show()
```

Distribution plots help understand:

shape<br>
spread<br>
relationship<br>
normality

Essential before feature scaling or modeling.
