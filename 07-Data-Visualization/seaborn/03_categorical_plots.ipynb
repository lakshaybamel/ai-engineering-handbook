# Categorical Plots in Seaborn

Categorical plots compare numerical values across categories.

Useful for:
- comparing groups
- feature analysis
- understanding distributions per category

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
```

```python
sns.barplot(x="day", y="total_bill", data=tips)

plt.title("Average Bill per Day")
plt.show()
```

```python
sns.barplot(x="day", y="total_bill", hue="sex", data=tips)

plt.title("Average Bill per Day by Gender")
plt.show()
```

```python
sns.countplot(x="day", data=tips)

plt.title("Customer Count per Day")
plt.show()
```

```python
sns.boxplot(x="day", y="total_bill", data=tips)

plt.title("Bill Distribution per Day")
plt.show()
```

```python
sns.violinplot(x="day", y="total_bill", data=tips)

plt.title("Distribution Shape per Day")
plt.show()
```

```python
sns.boxplot(x="day", y="total_bill", data=tips)
sns.stripplot(x="day", y="total_bill", data=tips, color="black", alpha=0.5)

plt.title("Box + Individual Observations")
plt.show()
```

```python
sns.catplot(x="day", y="total_bill", hue="sex", kind="bar", data=tips)

plt.show()
```

Categorical plots help compare groups and distributions.

Barplot → averages<br>
Boxplot → spread<br>
Violinplot → density<br>
Countplot → frequency
