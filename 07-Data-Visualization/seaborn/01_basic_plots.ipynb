# Seaborn Basic Plots

Seaborn is built on top of Matplotlib.
It provides better default styling and works directly with datasets.

We will learn:

- loading datasets
- simple plots
- automatic styling

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# seaborn comes with built-in datasets
tips = sns.load_dataset("tips")

tips.head()
```

```python
sns.scatterplot(x="total_bill", y="tip", data=tips)

plt.title("Total Bill vs Tip")
plt.show()
```

```python
sns.scatterplot(x="total_bill", y="tip", hue="sex", data=tips)

plt.title("Bill vs Tip by Gender")
plt.show()
```

```python
sns.set_style("darkgrid")

sns.scatterplot(x="total_bill", y="tip", data=tips)

plt.title("Dark Grid Style")
plt.show()
```

```python
sns.set_context("talk")

sns.scatterplot(x="total_bill", y="tip", data=tips)

plt.title("Talk Context (Bigger Plot)")
plt.show()
```

```python
sns.lineplot(x="size", y="total_bill", data=tips)

plt.title("Line Plot Example")
plt.show()
```

```python
sns.scatterplot(x="total_bill", y="tip", data=tips)

plt.savefig("seaborn_basic.png", dpi=300)
plt.show()
```
