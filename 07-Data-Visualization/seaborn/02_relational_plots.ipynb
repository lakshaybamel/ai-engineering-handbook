# Relational Plots in Seaborn

Relational plots show relationships between variables.

Seaborn provides:

- scatterplot()
- lineplot()
- relplot() (figure-level interface)

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
```

```python
sns.scatterplot(x="total_bill", y="tip", data=tips)

plt.title("Total Bill vs Tip")
plt.show()
```

```python
sns.scatterplot(x="total_bill", y="tip", hue="day", data=tips)

plt.title("Bill vs Tip by Day")
plt.show()
```

```python
sns.scatterplot(x="total_bill", y="tip", hue="day", size="size", data=tips)

plt.title("Scatter with Size Encoding")
plt.show()
```

```python
sns.lineplot(x="size", y="total_bill", data=tips)

plt.title("Party Size vs Total Bill")
plt.show()
```

```python
sns.relplot(x="total_bill", y="tip", hue="day", col="time", data=tips)

plt.show()
```

```python
sns.scatterplot(x="total_bill", y="tip", style="sex", hue="sex", data=tips)

plt.title("Different Markers by Gender")
plt.show()
```

```python
fmri = sns.load_dataset("fmri")

sns.lineplot(x="timepoint", y="signal", hue="region", data=fmri)

plt.title("Brain Signal Over Time")
plt.show()
```

Relational plots help analyze relationships, trends, and patterns.

Key features:<br>
hue → category<br>
size → magnitude<br>
style → marker type
